# Extracting Features from the Raw datsaet

storing No transaction Fail record from the dataset

```python
No_Fail=[]
for i in range (0,143):
  s=0
  for j in range(0,623):
        if(final_df['customerEmail'][i]==trans_df['customerEmail'][j]):
          s += trans_df['transactionFailed'][j]
  No_Fail.append(s)
final_df['No_Fail'] = No_Fail
```

Total Number of transactions

```python
total_amt = []
for i in range(0,143):
    s=0
    for j in range(0,623):
        if(final_df['customerEmail'][i]==trans_df['customerEmail'][j]):
            s += trans_df['transactionAmount'][j]
    total_amt.append(s)
final_df['Total_amt'] = total_amt
```

Payment registration failed report

```python
payRegFail = []
for i in range(0,143):
    s=0
    for j in range(0,623):
        if(final_df['customerEmail'][i]==trans_df['customerEmail'][j]):
            s += trans_df['paymentMethodRegistrationFailure'][j]
    payRegFail.append(s)
final_df['PaymentRegFail'] = payRegFail
```

```python
def col_make(column_name,category):
    array = []
    for i in range(0,143):
        s=0
        for j in range(0,623):
            if(final_df['customerEmail'][i]==trans_df['customerEmail'][j]):
                if trans_df[column_name][j]==category:
                    s+=1
        array.append(s)
    return array
```

creating extra columns in the final df for better prediction

```python
PaypalPayments = col_make('paymentMethodType','paypal')
ApplePayments = col_make('paymentMethodType','apple pay')
BitcoinPayments = col_make('paymentMethodType','bitcoin')
CardPayments = col_make('paymentMethodType','card')
OrdersFulfilled = col_make('orderState','fulfilled')
OrdersFailed =  col_make('orderState','failed')
OrdersPending = col_make('orderState','pending')
JCB_16 = col_make('paymentMethodProvider','JCB 16 digit')
AmericanExp = col_make('paymentMethodProvider','American Express')
VISA_16 =  col_make('paymentMethodProvider','VISA 16 digit')
Discover =  col_make('paymentMethodProvider','Discover')
Voyager = col_make('paymentMethodProvider','Voyager')
VISA_13 = col_make('paymentMethodProvider','VISA 13 digit')
Maestro = col_make('paymentMethodProvider','Maestro')
Mastercard = col_make('paymentMethodProvider','Mastercard')
DC_CB =col_make('paymentMethodProvider','Diners Club / Carte Blanche')
JCB_15= col_make('paymentMethodProvider','JCB 15 digit')
```

```python
final_df['PaypalPayments']= PaypalPayments
final_df['ApplePayments']= ApplePayments
final_df['CardPayments']= CardPayments
final_df['BitcoinPayments']= BitcoinPayments
final_df['OrdersFulfilled'] = OrdersFulfilled
final_df['OrdersPending'] = OrdersPending
final_df['OrdersFailed'] = OrdersFailed
final_df['JCB_16'] = JCB_16
final_df['AmericanExp'] = AmericanExp
final_df['VISA_16'] = VISA_16
final_df['Discover'] = Discover
final_df['Voyager'] = Voyager
final_df['VISA_13'] = VISA_13
final_df['Maestro'] = Maestro
final_df['Mastercard'] = Mastercard
final_df['DC_CB'] = DC_CB
final_df['JCB_15'] = JCB_15
```

Based on extracted features, the transaction fail order, duplicate ID, and duplicate address associted with the data are also segregated into different col

```python
Trns_fail_order_fulfilled = []
for i in range(0,143):
    s=0
    for j in range(0,623):
        if(final_df['customerEmail'][i]==trans_df['customerEmail'][j]):
            if (trans_df['orderState'][j]=='fulfilled') & (trans_df['transactionFailed'][j]==1):
                s+=1
    Trns_fail_order_fulfilled.append(s)
final_df['Trns_fail_order_fulfilled'] = Trns_fail_order_fulfilled
```

```python
Duplicate_IP = []
for i in range(0,143):
    s=0
    for j in range(0,143):
        if(final_df['customerIPAddress'][i]==final_df['customerIPAddress'][j]):
            s+=1
    s-=1
    Duplicate_IP.append(s)
final_df['Duplicate_IP'] = Duplicate_IP
```

```python
Duplicate_Address = []
for i in range(0,143):
    s=0
    for j in range(0,143):
        if(final_df['customerBillingAddress'][i]==final_df['customerBillingAddress'][j]):
            s+=1
    s-=1
    Duplicate_Address.append(s)
final_df['Duplicate_Address']=Duplicate_Address
```

```python
final_df = pd.get_dummies(final_df,columns=['Fraud'],prefix=['Fraud'],drop_first=True)
final_df = final_df.rename(columns={"Fraud_True": "Fraud"})
```

the Final dataset

```python
final_df.head()
```

Creating target variable

```python
x = final_df.drop(['customerEmail','customerPhone', 'customerDevice', 'customerIPAddress',
       'customerBillingAddress','Fraud'], axis=1)
y = final_df['Fraud']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

# Compiling data in Machine Learning Models

Logistic Regression

```python
Lr = LogisticRegression(solver='liblinear')
Lr.fit(x_train,y_train)
y_pred_lr= Lr.predict(x_test)
print(classification_report(y_test, y_pred_lr))
sns.heatmap(confusion_matrix(y_test, y_pred_lr), annot = True)
plt.show()
```

Random Forest

```python
model_rf = RandomForestClassifier()
model_rf = model_rf.fit(x_train, y_train)
y_pred_rd = model_rf.predict(x_test)
print(classification_report(y_test, y_pred_rd))
sns.heatmap(confusion_matrix(y_test, y_pred_rd), annot = True)
```

Prediction result

```python
accuracy, precision, recall, roc = [],[],[],[]
predictions = [y_pred_rd,y_pred_lr]


for i in range (len(predictions)):
    accuracy.append(accuracy_score(y_test, predictions[i]))
    precision.append(precision_score(y_test, predictions[i]))
    recall.append(recall_score(y_test, predictions[i]))
    roc.append(roc_auc_score(y_test, predictions[i]))

compare = pd.DataFrame ({'Accuracy': accuracy, 'Precision': precision, 'Recall': recall, 'ROC_score':roc}, index=['rf','lr'])
compare.sort_values('Accuracy', ascending=False)
```
